```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img_path = "/content/j.png"
img = cv2.imread(img_path)
plt.imshow(img)
#
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3, 3))
```

```python
dilated_img = cv2.dilate(img, kernel, iterations=3)
eroded_img = cv2.erode(img, kernel, iterations=3)
```

```python
opening_img = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
closing_img = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
```

```python
plt.figure(figsize=(10, 15))

plt.subplot(5, 1, 1)
plt.title("Original Image")
plt.imshow(img)
plt.axis("off")

plt.subplot(5, 1, 2)
plt.title("Dilated Image")
plt.imshow(dilated_img)
plt.axis("off")

plt.subplot(5, 1, 3)
plt.title("Eroded Image")
plt.imshow(eroded_img)
plt.axis("off")

plt.subplot(5, 1, 4)
plt.title("Opening Image")
plt.imshow(opening_img, cmap="gray")
plt.axis("off")

plt.subplot(5, 1, 5)
plt.title("Closing Image")
plt.imshow(closing_img, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```

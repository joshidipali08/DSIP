```python

import numpy as np
import matplotlib.pyplot as plt

# Input signal
signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])

# Compute FFT
fft_result = np.fft.fft(signal)

# Compute magnitude and phase spectrum
magnitude_spectrum = np.abs(fft_result)
phase_spectrum = np.angle(fft_result)

# Compute IFFT to reconstruct the signal
reconstructed_signal = np.fft.ifft(fft_result)

# Display results
print("Original Signal:")
print(signal)

print("\nFFT Result:")
print(np.round(fft_result, 2))

print("\nMagnitude Spectrum:")
print(np.round(magnitude_spectrum, 2))

print("\nPhase Spectrum (radians):")
print(np.round(phase_spectrum, 2))

print("\nReconstructed Signal:")
print(np.round(reconstructed_signal.real, 2))

# Plot all results
n = np.arange(len(signal))

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.stem(n, signal)
plt.title("Original Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2, 2, 2)
plt.stem(n, magnitude_spectrum)
plt.title("Magnitude Spectrum")
plt.xlabel("Frequency Index")
plt.ylabel("Magnitude")
plt.grid(True)

plt.subplot(2, 2, 3)
plt.stem(n, phase_spectrum)
plt.title("Phase Spectrum")
plt.xlabel("Frequency Index")
plt.ylabel("Phase (radians)")
plt.grid(True)

plt.subplot(2, 2, 4)
plt.stem(n, reconstructed_signal.real)
plt.title("Reconstructed Signal using IFFT")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
